## Curriculum Settings

Edit only the stage sizes or high-level run constants here. The stage construction and CLI argument plumbing are shared with the other notebooks.


```python
FORAGE_CONFIG = PROJECT_ROOT / "experiments" / "forage_curriculum.json"
experiment = workflows.load_jax_experiment(FORAGE_CONFIG)
RUN_DIR = PROJECT_ROOT / "runs" / "notebooks" / "forage_curriculum"
CHECKPOINT_DIR = RUN_DIR / "checkpoints"
MEDIA_DIR = RUN_DIR / "media"

STAGE_SIZES = range(4, 51, 4)
NUM_ENVS = 16
ROLLOUT_TILE_SIZE = workflows.NOTEBOOK_ROLLOUT_TILE_SIZE
ROLLOUT_POLICY_TEMPERATURE = workflows.notebook_rollout_policy_temperature(experiment.metadata)
ACTOR_VISION_RADIUS = 1
WRITE_BITS = 1
WRITE_WHILE_MOVING = True
WANDB_PROJECT = None
WANDB_ENTITY = None
WANDB_GROUP = "forage_curriculum_50x50"
WANDB_MODE = "online"
WANDB_VIDEO_MAX_FRAMES = 600
WANDB_VIDEO_STAGE_NAMES = workflows.FORAGE_WANDB_PREVIEW_STAGE_NAMES
WANDB_PROJECT = "cool-antz"

CURRICULUM_STAGES = workflows.build_forage_curriculum_stages(STAGE_SIZES)
NUM_STEPS = int(CURRICULUM_STAGES[0]["num_steps"])
GAMMA = float(CURRICULUM_STAGES[0]["gamma"])
GLOBAL_UPDATE_CAP = max(int(stage["global_update_cap"]) for stage in CURRICULUM_STAGES)
COMMON_ARGS = workflows.build_forage_common_args(
    CURRICULUM_STAGES, num_envs=NUM_ENVS, num_steps=NUM_STEPS,
    actor_vision_radius=ACTOR_VISION_RADIUS, write_bits=WRITE_BITS,
    gamma=GAMMA, write_while_moving=WRITE_WHILE_MOVING,
)
UPDATE_TIMESTEPS = workflows.update_timesteps(num_envs=NUM_ENVS, num_steps=NUM_STEPS)
PROFILE_SUMMARY = [(stage["name"], stage["global_update_cap"], stage["num_steps"], stage["gamma"]) for stage in CURRICULUM_STAGES]

{
    "stages": [stage["name"] for stage in CURRICULUM_STAGES],
    "max_updates_per_stage": GLOBAL_UPDATE_CAP,
    "budget_profile": PROFILE_SUMMARY,
    "wandb_preview_stages": WANDB_VIDEO_STAGE_NAMES,
}
```

## Train Curriculum Checkpoints


```python
forage_result = workflows.run_forage_curriculum(
    stages=CURRICULUM_STAGES,
    checkpoint_dir=CHECKPOINT_DIR,
    common_args=COMMON_ARGS,
    update_timesteps_per_stage=UPDATE_TIMESTEPS,
    global_update_cap=GLOBAL_UPDATE_CAP,
    train_main=jax_runner.main,
    wandb_project=WANDB_PROJECT,
    wandb_entity=WANDB_ENTITY,
    wandb_group=WANDB_GROUP,
    wandb_mode=WANDB_MODE,
    wandb_video_max_frames=WANDB_VIDEO_MAX_FRAMES,
    wandb_video_stage_names=WANDB_VIDEO_STAGE_NAMES,
    wandb_video_policy_temperature=ROLLOUT_POLICY_TEMPERATURE,
)
FINAL_CHECKPOINT_PATH = forage_result["final_checkpoint_path"]
forage_result
```
